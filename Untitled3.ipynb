```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("social_media_engagement_5000.csv")
df.head()
print(df.shape)
print(df.columns)
print(df.head())
print(df.tail())
df.info()
print(df.dtypes)
```

```python
likes = np.array([100, 200, 300, 400, 500])
print("Array:", likes)
print("Sum:", np.sum(likes))
print("Mean:", np.mean(likes))
print("Maximum:", np.max(likes))
print("Minimum:", np.min(likes))
print("Likes + 10:", likes + 10)
```

```python
print(df.isnull().sum())
df['age'] = df['age'].fillna(df['age'].median())
df['likes'] = df['likes'].fillna(df['likes'].median())
df['comments'] = df['comments'].fillna(df['comments'].median())
df['shares'] = df['shares'].fillna(df['shares'].median())
df['gender'] = df['gender'].fillna(df['gender'].mode()[0])
df['sentiment'] = df['sentiment'].fillna(df['sentiment'].mode()[0])
```

```python
print("Duplicate rows:", df.duplicated().sum())
df = df.drop_duplicates()
print("Duplicates after cleaning:", df.duplicated().sum())
```

```python
df['age'] = df['age'].astype(int)
df['likes'] = df['likes'].astype(int)
df['comments'] = df['comments'].astype(int)
df['shares'] = df['shares'].astype(int)
df['posted_at'] = pd.to_datetime(
    df['posted_at'],
    format='%d-%m-%Y'
)
df.info()
```

```python
df = df[
    (df['age'] >= 13) &
    (df['age'] <= 100) &
    (df['likes'] >= 0) &
    (df['comments'] >= 0) &
    (df['shares'] >= 0)
]
print(df[['age', 'likes', 'comments', 'shares']].describe())
```

```python
df['hashtag_count'] = df['hashtags'].apply(
    lambda x: len(str(x).split())
)
print(df[['hashtags', 'hashtag_count']].head())
```

```python
df['sentiment'] = df['sentiment'].str.strip().str.lower()
print(df['sentiment'].value_counts())
```

```python
df['engagement_score'] = (
    df['likes'] +
    df['comments'] +
    df['shares']
)
print(df[['likes', 'comments', 'shares',
          'engagement_score']].head())
```

```python
df['log_likes'] = np.log1p(df['likes'])
df['log_comments'] = np.log1p(df['comments'])
df['log_shares'] = np.log1p(df['shares'])
print(df[['log_likes', 'log_comments',
          'log_shares']].head())
```

```python
columns = [
    'likes',
    'comments',
    'shares',
    'watch_time_sec',
    'engagement_rate',
    'follower_count'
]
for col in columns:
    print("\nColumn:", col)
    print("Mean:", df[col].mean())
    print("Median:", df[col].median())
    print("Mode:", df[col].mode()[0])
```

```python
for col in columns:
    print("\nColumn:", col)
    print("Standard Deviation:", df[col].std())
    print("Variance:", df[col].var())
```

```python
for col in columns:
    print("\nColumn:", col)
    print("25th Percentile:", df[col].quantile(0.25))
    print("50th Percentile:", df[col].quantile(0.50))
    print("75th Percentile:", df[col].quantile(0.75))
    print("90th Percentile:", df[col].quantile(0.90))
```

```python
for col in columns:
    print("\nColumn:", col)
    print("Skewness:", df[col].skew())
    print("Kurtosis:", df[col].kurt())
```
